## Operadores relacionais

```
5 > 3
```

```
5 < 3
```

```
5 >= 5
```

```
5 <= 3
```

```
5 == 3    ### o símbolo == significa = ###
```

```
5 != 3    ### o símbolo != significa diferente ###
```

```
if 5 > 3:
  print('5 é maior que 3')
```

```
if 5 > 6:
  print('5 é maior que 6')
else:
  print('5 não é maior que 6')
```

```
if 5 > 4:
  print('5 é maior que 4')
else:
  print('5 não é maior que 4')
```

```
n = 3       ### mudar para n =4 e rodar ###
if n == 4:
  print('n é igual a 4')
else:
  if n == 3:
    print('n é igual a 3')
  else:
    print('n não é igual a 4 nem 3')
```

```
x = 2
y = 3                          ###  mudar para 4 #####
if (x > 1) and (y % 2 == 0):    ### o comando ### pega o resto da divisão por 2 ###  ### mudar para or ####
  print('x é maior que 1 e y é par')
else:
  print('x não é maior que 1 ou y não é par')
```
