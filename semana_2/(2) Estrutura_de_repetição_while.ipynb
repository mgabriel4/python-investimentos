# Estruturas de repetição - while

```
numero = 1
while numero < 6:
  print(numero)
  numero += 1    # +=1 é o incremento ##
print('---')
```

```
numero = 5
while numero > 0:
  print(numero)
  numero -= 1     ### decrementar -= -1 ####
```

```
soma = 0
numero = 1
while numero < 6:
  soma += numero
  numero += 1
print(soma)
```
