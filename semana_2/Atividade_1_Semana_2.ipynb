# Exercício 1

Ler 5 notas e informar a média


## For

```python
soma = 0

for i in range(5):
    nota = float(input(f"Digite a {i + 1}ª nota: "))
    soma += nota

media = soma / 5

print(f"A média das notas é: {media:.2f}")
```

## While

```python
soma = 0
contador = 1

while contador <= 5:
    nota = float(input(f"Digite a {contador}ª nota: "))
    soma += nota
    contador += 1

media = soma / 5

print(f"A média das notas é: {media:.2f}")
```

# Exercício 2

Imprimir a tabuada do número 3 (3 x 1 = 3 até 3 x 10 = 30)


## For

```python
for i in range(1, 11):
    resultado = 3 * i
    print(f"3 x {i} = {resultado}")
```

## While

```python
i = 1

while i <= 10:
    resultado = 3 * i
    print(f"3 x {i} = {resultado}")
    i += 1
```
