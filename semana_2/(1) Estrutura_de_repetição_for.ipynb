# Estruturas de repetição - for

```
print(1)
print(2)
print(3)
print(4)
print(5)
```

```
for numero in range(1, 6):  ### range(início, fim) inclui o início e exclui o fim.
  print(numero)
```

```
for numero in range(5, 0, -1):   ### -1 é para ele subtrair -1 a partir do 6 ###
  print(numero)
```

```
5 + 4 + 3 + 2 + 1
```

```
soma = 0
for numero in range(1, 6):
  soma = soma + numero
print(soma)
```

```
palavra = 'sorvete'
for letra in palavra:
    if letra == 'v':
        print('Achou a letra v')
```

```
for i in range(0,5):
  print(i)
  print('---')
  for j in range(0,3):
    print(j)
  print() ### para pular uma linha ###
```
