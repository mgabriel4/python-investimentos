## Dicionários

```
dicionario = {'Palmeiras': 12, 'Corinthians': 7, 'São Paulo': 6} ## armazenam dados ordenados em pares chave-valor###
```

```
dicionario['São Paulo']
```

```
dicionario['Santos'] = 8   #### acrescenta um elemento no dicionário ####
```

```
print(dicionario)
```

```
del(dicionario)['Corinthians']    ### del = apaga um elemento no dicionário ##
print(dicionario)
```

```
dicionario.items() ### item para visualizar o dicionário inteiro ###
```

```
dicionario.keys()    #### keys traz apenas as chaves sem os valores ###
```

```
dicionario.values()   ### traz os valores das chaves ####
```

```
dicionario2 = {'Flamengo': 8, 'Vasco': 4}
print(dicionario2)
```

```
dicionario.update(dicionario2)    ### update junta os valores de dois dicionários ##
print(dicionario)
```

```
dicionario.items()   ### retorna chave-valor ####
```

```
for i, j in dicionario.items():
  print(f'Time: {i}, número de títulos do brasileirão: {j}')  ##f para formatação de string (para pegar a chave-valor) ###
```

## Conjuntos (set)

```
times = {'Palmeiras', 'Corinthians', 'São Paulo', 'Palmeiras',
                'Corinthians', 'Santos', 'Santos', 'Santos'}  ## não ordenados e armazenam elementos únicos ###
```

```
print(times)
```

```
print(set(times))   ### set seleciona os elementos q não se repetem ###
```

```
c1 = {1,2,3,4,5}
c2 = {3,4,5,6,7}
c3 = c1.intersection(c2)
```

```
print(c3)
```

```
c1.difference(c2)   ### C1 - C2 ###
```

```
c2.difference(c1)   ### C2 - C1 ###
```

#**Matrizes**

```
import numpy as np
```

```
matriz = np.array([[2, 3, 1], [4, 5, 7]])
```

```
matriz
```

```
matriz.shape   ###shape é o comando para dizer o tipo da matriz ###
```

```
matriz[0]
```

```
matriz [1]
```

```
matriz[0][1]    #### pega o elemento da linha 0 e coluna 1 ####
```

```
matriz[1][2]
```

```
for i in range(matriz.shape[0]):  #### shape[0] conta as linhas, no caso a 1 e matriz[i] pega o conteúdo da linha 1
  print(matriz[i])
```

```
for i in range(matriz.shape[0]):   ### mostra a linha inteira ###
  print(matriz[i])
  for j in range(matriz.shape[1]):  ### esse for de j percorre cada elemento dessa linha
    print(matriz[i][j])
```
