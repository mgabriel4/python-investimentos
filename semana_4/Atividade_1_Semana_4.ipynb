b. Extract data from yfinancefor Ford from the 1st of January 2024. The ticker you need is ‘F’.

```python
import yfinance as yf

F = yf.download('F', start='2024-01-01')
print(F.head())
```

```python
F
```

c. Data Visualization

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

close = F['Close']
if isinstance(close, pd.DataFrame):
    close = close.iloc[:, 0]

plt.plot(F.index, close, linewidth=1.5, label='Close Price')
plt.title('Ford (F) - Closing Stock Price Since January 1, 2024')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

**Interpretação:** O gráfico de linhas mostra a evolução do preço de fechamento das ações da Ford desde 1º de janeiro de 2024, facilitando a identificação de períodos de valorização, queda e volatilidade dos preços ao longo do tempo.
