Practice extracting closing price data from the same source and for the same timeframe. Do this for Apple, Microsoft, Exxon, and British Petroleum. Use Yahoo Finance to find the tickers for these companies if you don’t already know them.

```python
import pandas as pd

acoes = ['AAPL', 'MSFT', 'XOM', 'BP']
data = pd.DataFrame()
inicio = '2026-01-01'
fim = '2026-09-10'
data = yf.download(acoes, start=inicio, end=fim)['Close']
```

```python
data
```

c. Data Visualization

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
sns.lineplot(data=data)
plt.xlabel('Data')
plt.ylabel('Preço de Fechamento ($)')
plt.title('Evolução dos Preços das Ações')
```
