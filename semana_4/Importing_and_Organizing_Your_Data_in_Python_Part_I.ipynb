```python
import yfinance as yf

PG = yf.download('PG', start='2026-01-01')
###PG (Procter & Gamble) ###

```

ticker é um código único usado para identificar uma ação, fundo ou outro ativo financeiro negociado em bolsa. No nosso caso, PG é o ticker da Procter & Gamble na Bolsa de Valores de Nova York (NYSE).

Date - Data do pregão.

Close - Preço de fechamento da ação no dia.

High - Maior preço alcançado pela ação no dia.

Low - Menor preço alcançado pela ação no dia.

Open - Preço de abertura da ação no dia.

Volume - Número total de ações negociadas no dia.

Visualização

```python
import matplotlib.pyplot as plt
# Baixar dados
tickers = ['PG']
data = yf.download(tickers, start='2026-01-01')['Close']

# Criar gráfico
plt.figure(figsize=(12,6))
for ticker in tickers:
    plt.plot(data.index, data[ticker], label=ticker)

plt.xlabel('Data')
plt.ylabel('Preço de Fechamento ($)')
plt.title('Evolução dos Preços das Ações')
plt.legend()
plt.grid(True)
plt.show()
```
