```python
tickers = ['PG', 'MSFT', 'T', 'F', 'GE'] #lista de ações
```

```python
new_data = pd.DataFrame()
```

```python
inicio = '2026-01-01'
fim = '2026-09-10'
```

```python
new_data = yf.download(tickers, start=inicio, end=fim)['Close']
```

```python
new_data
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

tickers = ['PG', 'MSFT', 'T', 'F', 'GE']
for i in tickers:
  plt.plot(new_data.index, new_data[i], label =i)


plt.xlabel('Data')
plt.ylabel('Preço de Fechamento ($)')
plt.title('Evolução dos Preços das Ações')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

tickers = ['PG', 'MSFT', 'T', 'F', 'GE']
data = yf.download(tickers, start=inicio, end=fim)['Close']

plt.figure(figsize=(12,6))
sns.lineplot(data=data)
plt.xlabel('Data')
plt.ylabel('Preço de Fechamento ($)')
plt.title('Evolução dos Preços das Ações')

# na biblioteca seaborn, não é necessário iterar (fazer o for) sobre cada ação, pois a função lineplot já faz isso automaticamente.
```
